### 3 · Check the tools the course needs

The course uses three **libraries** (add-ons to Python): `pandas` for tables, `matplotlib` for charts, and
`requests` for getting data from the internet. Anaconda and Google Colab include all three. This cell
checks that they are there.

```python
import sys                                        # sys: information about the running Python
print("Python version:", sys.version.split()[0])  # show the version number, for example 3.12.4

missing = []                                      # we collect here any library that is not installed
for name in ["pandas", "matplotlib", "requests"]: # the three libraries the course uses
    try:
        library = __import__(name)                # try to load the library by its name
        print(f"[OK]      {name} {library.__version__}")   # it loaded: show its version
    except ImportError:                           # it could not be loaded
        missing.append(name)                      # remember it
        print(f"[MISSING] {name}")

if missing:                                       # at least one library is missing
    print("\nSome tools are missing. Start Jupyter Notebook from Anaconda Navigator and run this cell again.")
else:
    print("\nAll the tools the course needs are installed.")
```

### 4 · Open the course dataset

The course comes with a dataset: 334 records about works by Vincent van Gogh, from Europeana. It is a
**CSV file**, a table saved as plain text, in the `datasets` folder. This cell finds the file and opens it
as a table.

*Have your own dataset?* Copy your CSV file into the `datasets` folder, write its name in `DATASET_FILE`
below, and run this cell and section 6 again.

```python
from pathlib import Path        # Path: a safe way to work with folders and file names
import pandas as pd             # pandas: the library for tables (by convention called pd)

DATASET_FILE = "van_gogh_combined.csv"   # the file to open (you can put your own file name here)

def find_dataset(name):
    """Return the path to a file in the course's datasets folder, wherever the notebook is run from."""
    for folder in [Path("datasets"), Path("../datasets")]:   # the folder is here, or one level up
        if (folder / name).exists():                          # is the file in this folder?
            return folder / name                              # yes: return its path
    raise FileNotFoundError(
        f"Could not find {name}. Open this notebook from inside the course folder, and make sure "
        "the course was unpacked (on Windows: right-click the ZIP > Extract All).")

path = find_dataset(DATASET_FILE)        # find the file
df = pd.read_csv(path)                   # read the CSV file into a table; we call the table df
print("Opened:", path)
print("Rows (records):", len(df), "| Columns (fields):", len(df.columns))
print("Column names:", list(df.columns))
```

### 5 · Draw a small chart

`value_counts()` counts how often each value appears in a column. Here we count the records of each
**type** (images, texts...) and draw the counts as a bar chart.

```python
import matplotlib.pyplot as plt          # the plotting library (by convention called plt)

counts = df["type"].value_counts()       # how many records of each type
print(counts)                            # show the numbers

counts.plot(kind="bar", color="#2E75B6") # draw them as a bar chart
plt.title("Records in the course dataset, by type")   # chart title
plt.ylabel("Number of records")          # label of the vertical axis
plt.xticks(rotation=0)                   # keep the labels horizontal
plt.tight_layout()                       # make sure nothing is cut off
plt.show()                               # display the chart
```
