## Step 1 · Look back over the course

Across the course you:

- understood **open research in the humanities**, and why FAIR and CARE belong together (Module 1);
- **documented** your project as you went and put it under **version control** on GitHub (Module 2);
- **found, assessed, and obtained** open heritage data, and learned why the same entity can have several
 names (Module 3);
- used **AI as a checked, documented tool** to filter your data (Module 4);
- **cleaned, analysed, visualised, and interpreted** your data carefully, stating their limits (Module 5);
- *(optionally)* **enriched** it by linking entities to Wikidata identifiers (Module 6);
- **audited** your project for reproducibility with the TIER protocol and a peer test (Module 7);
- **published** your created dataset openly with a DOI, and checked it against FAIR (Module 8).

Two threads ran through all of it: **authority control** (the same thing written many ways, made reliable
with stable identifiers) and **documenting every decision**, so that someone else can follow, check, and
reuse your work.

**What the course did not cover.** This is an introduction. Larger topics, such as advanced text and
network analysis, linked-data modelling, long-term preservation, and domain-specific ethics, are natural
next steps. The habits you practised here will serve you in all of them.

**Goal:** after reading, you can answer these questions.

1. Which two threads ran through the whole course?
2. In which order do the main stages of the course come?
3. Why is documentation needed at every stage, and not just at the end?

<details>
<summary>Check your answers</summary>

1. Authority control (stable identifiers for things written in many ways) and documenting every decision.
2. Openness, documentation, obtaining data, documenting AI, analysis, (enhancement), reproducibility, publishing.
3. Decisions are hard to reconstruct later, and every stage changes the data; only notes made along the way let others follow and check the work.

</details>

### 3.1 · Check your project, module by module

Your project now holds what you produced in each module. Open your repository (on GitHub, or the folder on
your computer) and tick off each piece:

| Module | What you produced | Where in your project |
|---|---|---|
| 1 | Reflection on access, provenance, rights, uncertainty, and reuse | your reflection log |
| 2 | Workflow map, README, folder structure, decision log | `docs/workflow_map.md`, `README.md`, the folders, `docs/decision_log.md` |
| 2 and 5 | Version history, or a change log | your commits on GitHub, or `docs/change_log.md` |
| 3 | Dataset profile with suitability checklist | `docs/dataset_profile.md` |
| 4 | AI-use statement and verification log | `docs/ai_use_and_verification.md` |
| 5 | Transformation record and documented notebook | `docs/transformation_record_....md`, `outputs/` |
| 6 | Enhancement table (optional module) | `docs/enhancement_table.md` |
| 7 | Reproducibility report, revision log, unresolved limits, peer test | `docs/reproducibility_report.md`, `docs/revision_log.md`, `docs/unresolved_limits.md`, your test template |
| 8 | Metadata, citation, README, licence or access statement, release decision | `publication/`, `docs/release_decision.md` |

Then read the README as if you had never seen the project: does it say what the project is, where the data
comes from (with licence and access date), and in which order to run the notebooks? If you published, is
the DOI linked in the README and in `CITATION.cff`?

## Step 6 · Working with your own dataset

**Using the course dataset (Van Gogh)?** You have finished the module. Look at the further reading below if you want to go deeper, then continue with the next module.

**Using your own dataset?** Do the general task below. On this track you work independently: use the code in Step 4 as a starting point, adapt it to your data, and use an AI assistant to help with the code where you need it (vibe coding). Always read and test what the AI gives you, and record how you used it.

**General task: finish your project.** Go through Step 3 with your own repository. Make sure the README tells
the story of your dataset from the source to the publication, and that every decision you made is
recorded somewhere a reader can find it.

**Self-reflection.** Open your reflection log (template: [`reflection-log-template.md`](../00-introduction/reflection-log-template.md)) and write a few sentences under *Course Wrap-up*:

- How did working with your own dataset go, overall? What went well, and what was harder than you expected?
- What did your data allow you to answer, and what did it not?
- Which open-research practice will you keep using, and which do you still find difficult?
- What would you tell someone who is about to start this course with their own dataset?
