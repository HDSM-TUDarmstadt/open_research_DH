## Step 1 · Read the introduction (this module has no video)

**The cycle of Digital Humanities research.** A typical cycle runs: *obtain > clean and transform >
analyse > visualise > interpret > document > share*. You obtained your data in Module 3 and filtered it in
Module 4. Now you turn it into an answer.

- **Cleaning** means small, honest tidying before you count: consistent spelling, no stray spaces, no
 empty groups, one date format. It never means inventing or changing values. Always record what you did.
- **Make variables that answer your question**, for example counts per institution, per country, or per
 decade.
- **Visualise simply.** One clear, labelled bar chart that shows the data well is better than a
 complicated chart with too much in it. Remember that a chart is an interpretation too: choose the type
 deliberately (see Johanna Drucker in the further reading).
- **Write a narrative.** Describe the distribution (where there is more, where less), point out what
 stands out, and state at least one limitation. Do not claim more than the data can support.

**Goal:** after reading, you can answer these questions.

### Configuration

```python
# Short name of your dataset (must match the earlier notebooks).
DATASET_PREFIX = "van_gogh"

# The column to count and chart. Try "country", "type", or "year" too.
GROUP_BY = "institution"

# The question this chart answers, shown as the chart title. Edit for your own study.
RESEARCH_QUESTION = "How are the works distributed across the holding institutions?"

# Your own project from Module 2. Results are also copied there, if the folder exists.
from pathlib import Path
PROJECT_FOLDER = Path.home() / "Documents" / "GitHub" / "open-project-2026"

print("Grouping by:", GROUP_BY)
```

### 1 · Load the filtered data (and a word on cleaning)

Cleaning here means small, honest tidying (consistent text, no blank groups) *before* counting. We do not
invent or change values. The helpers below find the shared `datasets/` folder wherever you run from.

```python
from pathlib import Path            # cross-platform file paths
import pandas as pd                   # tables
import matplotlib.pyplot as plt       # plotting (charts)

FILTERED_FILE = f"{DATASET_PREFIX}_filtered.csv"   # filename derived from the prefix

def find_dataset(name):
    """Return the path to `name`, looking in ./datasets and ../datasets."""
    for base in [Path("datasets"), Path("../datasets")]:   # the folder is here, or one level up
        if (base / name).exists():
            return base / name
    raise FileNotFoundError(f"Could not find {name} in datasets/ or ../datasets/")

def load_filtered():
    """Load the filtered dataset from Module 4 (falls back to earlier files if needed)."""
    for name in [FILTERED_FILE, "van_gogh_filtered.csv", "van_gogh_combined.csv"]:
        try:
            path = find_dataset(name); print("Loaded", path)
            return pd.read_csv(path), name
        except FileNotFoundError:
            continue                                         # not there: try the next name
    raise FileNotFoundError("Run Module 4 first.")

PROC_DIR = Path("data/processed"); PROC_DIR.mkdir(parents=True, exist_ok=True)   # outputs go here
STEPS = []                                                   # the transformation record: one entry per step

df, INPUT_NAME = load_filtered()                             # load the table
STEPS.append(("Load", f"{INPUT_NAME}", len(df)))

if GROUP_BY not in df.columns:                               # the column must exist
    raise KeyError(f"Column {GROUP_BY!r} not found. Available columns: {list(df.columns)}")

missing_share = df[GROUP_BY].isna().mean()                   # fraction of empty cells (0 to 1)
if missing_share > 0.5:
    print(f"WARNING: {GROUP_BY!r} is {missing_share:.0%} empty, so the chart only shows the records that have a value.")

df = df[df[GROUP_BY].notna()].copy()                         # keep only rows that have a value
STEPS.append((f"Drop rows with no {GROUP_BY}", f"{missing_share:.0%} of rows were empty", len(df)))
if pd.api.types.is_float_dtype(df[GROUP_BY]) and (df[GROUP_BY] % 1 == 0).all():
    df[GROUP_BY] = df[GROUP_BY].astype(int)                  # show 1888 rather than 1888.0
    STEPS.append((f"Write {GROUP_BY} as whole numbers", "1888.0 becomes 1888", len(df)))
df[GROUP_BY] = df[GROUP_BY].astype(str).str.strip()          # light cleaning: trim stray spaces
df = df[df[GROUP_BY] != ""]                                   # drop values that were only spaces
STEPS.append(("Trim spaces, drop blank values", "no values invented or changed", len(df)))

print("Records with a value:", len(df), "| grouping by:", GROUP_BY)
```

### 2 · Count and visualise

We count how many records fall into each group, then draw a **horizontal** bar chart (horizontal bars read
better for long labels such as institution names).

#### Your turn: count and chart

**Task:** count how many records each institution has, and draw the counts as a horizontal bar chart with a title and labelled axes.

**Hint:** `df[GROUP_BY].value_counts()` counts; `.sort_values().plot(kind="barh")` draws the bars with the largest on top; `ax.set_title(...)` and `ax.set_xlabel(...)` add the labels.

**You should get:** 10 institutions; the Catholic University of Leuven has by far the longest bar (18).

Try it in the empty cell below. Then unfold the **solution** cell under it, compare, and run the solution
cell too, because the next steps use it.

```python
# Your attempt: write your code here, then press Shift + Enter
```

```python
#@title Solution (unfold to see the code)
# value_counts() tallies rows per distinct value; this is the analysis in one line.
counts = df[GROUP_BY].value_counts()
print(counts)

ordered = counts.sort_values()                    # ascending, so the largest bar ends up on top
ax = ordered.plot(kind="barh", figsize=(9, 5), color="#2E75B6")   # horizontal bar chart

ax.set_title(RESEARCH_QUESTION, fontsize=11)      # chart title = your question
ax.set_xlabel("Number of works")                  # x-axis label
ax.set_ylabel(GROUP_BY.capitalize())              # y-axis label (the group name)

# Whole-number ticks only (0, 2, 4, ...): counts are integers, so avoid 17.5-style ticks.
top = int(ordered.max())                          # the tallest bar's value
step = max(1, top // 10)                          # about ten ticks, at least 1 apart
ax.set_xticks(range(0, top + 1, step))            # place the ticks

plt.tight_layout()                                # stop labels being clipped
plt.savefig(PROC_DIR / f"works_per_{GROUP_BY}.png", dpi=150)   # save the chart as an image
plt.show()                                         # display it here
```

```python
# Save the counted table too, so the numbers behind the chart are reproducible.
counts.rename("count").to_csv(PROC_DIR / f"works_per_{GROUP_BY}.csv")
print("Saved table + chart to", PROC_DIR)
```

#### The transformation record

This is the result of Module 5. Every step between loading the data and the chart is listed with the
number of rows after it, so anyone can check what happened to the data. Open the file afterwards and add
your interpretation (section 3).

```python
import datetime

STEPS.append((f"Count records per {GROUP_BY}", f"{len(counts)} groups", len(df)))
step_lines = "\n".join(f"| {i} | {step} | {note} | {rows} |" for i, (step, note, rows) in enumerate(STEPS, 1))
record = f"""# Transformation record: analysis by {GROUP_BY}

*Created {datetime.date.today().isoformat()} by the course notebook 05_data_analysis.*

**Question:** {RESEARCH_QUESTION}

| # | Step | Note | Rows after the step |
|---|---|---|---|
{step_lines}

**Outputs:** `works_per_{GROUP_BY}.png` (chart) and `works_per_{GROUP_BY}.csv` (the numbers behind it).

**Interpretation and limitations:** (fill in your three or four sentences from section 3)
"""
record_path = PROC_DIR / f"transformation_record_{GROUP_BY}.md"
record_path.write_text(record, encoding="utf-8")
print(record)
print("Saved:", record_path)
```

Your project from Module 2 collects everything you produce, so that in Module 7 you can audit it as one

```python
import shutil, sys

if "google.colab" in sys.modules:                       # on Colab the project lives in /content
    PROJECT_FOLDER = Path("/content") / PROJECT_FOLDER.name

def to_project(file, subfolder):
    """Copy one result file into your project, for example into data/processed or docs."""
    file = Path(file)
    if not PROJECT_FOLDER.exists():                     # no project: leave the file where it is
        print(f"[NOT COPIED] {file.name}: project folder not found. The file is here: {file.resolve()}")
        return
    target = PROJECT_FOLDER / subfolder                 # for example .../open-project-2026/docs
    target.mkdir(parents=True, exist_ok=True)           # create the folder if needed
    shutil.copy(file, target / file.name)               # copy (the original stays where it is)
    print(f"[COPIED]     {file.name} -> {target}")

to_project(PROC_DIR / f"works_per_{GROUP_BY}.png", "outputs")       # the chart
to_project(PROC_DIR / f"works_per_{GROUP_BY}.csv", "outputs")       # the numbers behind it
to_project(record_path, "docs")                                       # the transformation record
```
