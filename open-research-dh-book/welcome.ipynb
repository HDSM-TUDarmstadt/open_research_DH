Do you always trust research results when you read a paper or look at a dataset? Do you ask what was done
to build the dataset, and what sequence of steps led to the conclusions? If so, you are already thinking
about open research. In this course you practise it with a running example: works by Vincent van Gogh,
collected from Europeana, which you obtain, filter, analyse, enrich, check, and publish openly.

**Who it is for.** Students and early-career researchers in the humanities, researchers who want to update
their open-research skills, anyone working with cultural-heritage data, and teachers who want to reuse open
materials. Every line of code is explained, and the setup is described click by click.

## How every module works

<div class="steps">

1. **Watch the video.** After watching, you can answer a few questions about it (answers included).
2. **The practical task.** First the goal: what you will learn to do.
3. **Tutorial: step by step.** Written instructions that spell out every step (with a tutorial video at the
 top, where a module has one).
4. **Run the code**, right on the same page. It starts with a short abstract of what the code does.
5. **Checklist.** What you can do now, plus tasks to repeat and simple questions to check yourself.
6. **Working with your own dataset.** An optional, independent track, ending with a short self-reflection.

Each module ends with **Further Reading and Resources**.

**Choose your route.** The course adapts to what you already know and what you have:

- **Setup:** install the tools on your computer, or use Google Colab in the browser (Module 0).
- **Dataset:** use the course's prepared Van Gogh dataset (every notebook works with it out of the box), or
 your own dataset, with the general task in Step 6 of every module. On the own-dataset track you work
 independently (an AI assistant can help with code), and record how it went in your
 [reflection log](00-introduction/reflection-log-template.md).
- **Keeping track of changes:** GitHub (recommended), or a simple change log (Module 2).
- **Prior knowledge:** if you already know a tool, Step 3 of each module starts with a *Choose your
 starting point* note that tells you what you can skip.

**Try first, then check.** The code contains *Your turn* tasks: try the task yourself in an empty
cell, then unfold the worked solution underneath to compare.

</div>

## One page per module

Each module is **one notebook**: you read it from top to bottom, and the code you run sits in Step 4 of
the same page. You can use it in three ways:

- **Read it here**, on this website. Click the rocket icon at the top of a page and choose **Colab** to run
 the code in your browser, with nothing to install.
- **Run it on your computer** in Jupyter Notebook. Module 0 explains how to install it and how to download
 the course.
- **Look at it on GitHub**, where every notebook is shown as a page too.
