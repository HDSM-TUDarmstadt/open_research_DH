### Part 1 · Obtaining a dataset: an overview of the options

**Where heritage data lives.** There are two kinds of source:

- **Museums, libraries and archives** that publish their own collections, for example the
 [Metropolitan Museum of Art](https://www.metmuseum.org/hubs/open-access), the
 [Smithsonian](https://www.si.edu/openaccess), and the [Rijksmuseum](https://data.rijksmuseum.nl/).
- **Aggregators** that bring many institutions together, for example
 [Europeana](https://www.europeana.eu) and the [European Data portal](https://data.europa.eu/).

A longer, checked list is in the [Open Heritage Data source catalogue](open-heritage-data-sources.ipynb).

**A checklist for any source.** Open the Met's open-access page and one object record, and ask:

1. How easy is it to get the data at all?
2. In what form? A **direct download** (CSV, JSON or ZIP), an **open API** you can query without
 registering, an **API that needs a key**, or **preview only** (you can look, but not take the data)?
3. Which **fields** does one record have (title, creator, date, medium, place...), and how complete are
 they? A plain record with twenty well-filled fields is more useful for research than a beautiful image
 with three.
4. Which **licence** applies? CC0, CC BY, or rights that vary from item to item?

**CSV or API?** A **CSV download** is simple: no code, and fine for a small dataset. But it is a fixed
snapshot: you cannot filter it while downloading, and it does not update. An **API** gives you current
data, lets you filter as you go (for example only one creator), and, because it is code, can be run again
by anyone. The cost is that it needs some code, and sometimes a key. Base your decision on your research needs.

### Where the data lives, and small helper functions

These lines set up **pandas** (a table toolkit) and two helper functions that find the shared
`datasets/` folder no matter which folder you launched the notebook from.

```python
from pathlib import Path            # Path: a safe, cross-platform way to handle file paths
import pandas as pd                   # pandas: the library we use for tables (DataFrames)
import requests                       # requests: fetches data over the internet (the API)

# Filenames are built from DATASET_PREFIX so renaming the creator renames the files too.
COMBINED_FILE = f"{DATASET_PREFIX}_combined.csv"   # e.g. "van_gogh_combined.csv"

# Names we accept when looking for an already-saved dataset (yours first, then course samples).
KNOWN_DATASETS = [COMBINED_FILE, "van_gogh_combined.csv", "van_gogh_europeana.csv"]

def find_dataset(name):
    """Return the path to a data file called `name`, looking in ./datasets and ../datasets."""
    for base in [Path("datasets"), Path("../datasets")]:   # check both likely locations
        candidate = base / name                            # build the full path, e.g. datasets/x.csv
        if candidate.exists():                             # does that file exist on disk?
            return candidate                               # yes: hand back the path
    raise FileNotFoundError(f"Could not find {name} in datasets/ or ../datasets/")

def datasets_dir():
    """Return the datasets/ folder to SAVE into (so new files sit beside any existing data)."""
    for base in [Path("datasets"), Path("../datasets")]:   # prefer a folder that already has data
        if any((base / n).exists() for n in KNOWN_DATASETS):
            return base
    Path("datasets").mkdir(parents=True, exist_ok=True)    # none yet: create ./datasets
    return Path("datasets")

# A place to keep the untouched "raw" pull, separate from cleaned data (good practice).
RAW_DIR = Path("data/raw")           # the folder path
RAW_DIR.mkdir(parents=True, exist_ok=True)   # create it if it does not exist (no error if it does)

print("Helpers ready. Files will be saved with the prefix:", DATASET_PREFIX)
```

### 1 · The name-variation problem (authority control)

The *same* creator returns very different results depending on the exact string you search. For Van Gogh
on Europeana:

| Query | Records |
|---|---|
| `CREATOR:"Vincent van Gogh"` | 24 |
| `CREATOR:"Gogh, Vincent van"` | 309 |
| `who:"Vincent van Gogh"` | 354 |

So we search a **list of variants** and combine them, then **de-duplicate**, because one record can match
several variants. This problem is addressed in **Module 6**, where we aim to give each entity a stable
Wikidata **Q-ID**.

> **Prompt you could use if you are querying a LLM:** *List plausible catalogue name variants for a creator: natural form,
> "surname, first name", with and without middle names, and common transliterations, so I can search each
> and combine.*

#### The functions that fetch the data
Two functions do the work: one fetches every record for a **single** name variant, the other loops over
**all** variants and combines them. Every line is commented.

```python
def first_value(value):
    """Europeana returns some fields as a list; take the first item, or pass a plain value through."""
    if isinstance(value, list):          # is this field a list (e.g. several titles)?
        return value[0] if value else None   # yes: take the first, or None if the list is empty
    return value                         # not a list: return it unchanged

def fetch_variant(variant, api_key):
    """Fetch every record whose CREATOR exactly matches one name `variant`."""
    url = "https://api.europeana.eu/record/v2/search.json"   # the Europeana search endpoint
    collected = []                       # a list we will fill with one dict per record
    cursor = "*"                         # "*" tells Europeana to start at the first page
    pages = 0                            # how many pages we have fetched so far

    # Keep asking for pages until there are none left, or we hit the safety cap.
    while cursor and pages < MAX_PAGES:
        params = {                       # the query settings sent with the request
            "wskey": api_key,            # your API key
            "query": "*:*",              # match everything...
            "qf": f'CREATOR:"{variant}"',# ...then filter to this exact creator string
            "rows": ROWS_PER_PAGE,       # how many records per page
            "cursor": cursor,            # where this page starts (Europeana gives us the next one)
        }
        response = requests.get(url, params=params, timeout=30)   # send the request (wait up to 30s)
        response.raise_for_status()      # stop with a clear error if the request failed (e.g. bad key)
        data = response.json()           # turn the JSON reply into a Python dictionary

        for item in data.get("items", []):        # loop over the records on this page
            collected.append({                     # keep only the fields we care about
                "id":          item.get("id"),                          # unique record id
                "title":       first_value(item.get("title")),          # work title
                "type":        first_value(item.get("type")),           # e.g. IMAGE, TEXT
                "institution": first_value(item.get("dataProvider")),   # who provided the record
                "country":     first_value(item.get("country")),        # provider country
                "year":        first_value(item.get("year")),           # year, if present
                "matched_variant": variant,                             # which name form matched
            })

        cursor = data.get("nextCursor")  # Europeana hands back the cursor for the next page (or nothing)
        pages += 1                       # count this page

    return collected                     # the list of record-dicts for this variant

def fetch_all_variants(variants, api_key):
    """Fetch every variant, combine the results, and remove duplicate records."""
    rows = []                            # collect records from all variants here
    for variant in variants:             # go through each name form in turn
        got = fetch_variant(variant, api_key)      # fetch this variant
        print(f"  {variant!r}: {len(got)} records")# report how many it returned
        rows += got                      # add them to the running list

    df = pd.DataFrame(rows)              # turn the list of dicts into a table (DataFrame)
    before = len(df)                     # how many rows before de-duplicating
    df = df.drop_duplicates(subset="id").reset_index(drop=True)  # keep one row per unique id
    print(f"Combined {before} rows -> {len(df)} unique records")  # report the de-duplication
    return df

def load_sample():
    """When there is no key, load the combined result shipped with the course."""
    # Try the file named for your prefix first, then fall back to the Van Gogh sample.
    for name in [COMBINED_FILE, "van_gogh_combined.csv"]:
        try:
            path = find_dataset(name)    # look for the file
            print("Loaded sample:", path)
            return pd.read_csv(path)     # read the CSV into a DataFrame and return it
        except FileNotFoundError:
            continue                     # not there: try the next name
    raise FileNotFoundError("No sample dataset found in datasets/. Add a key to fetch live data.")
```

### 5 · Write the dataset profile

The **dataset profile** is the result of Module 3. This cell writes a first draft from the data itself:
source and provenance, scope, every field with its gaps, rights, limitations, and a **suitability
checklist**. Open `outputs/dataset_profile.md` afterwards and complete every line marked *(fill in)*:
those parts need your judgement.

```python
import datetime

today = datetime.date.today().isoformat()
providers = df[ENTITY_COLUMN].value_counts()                       # records per provider
top_share = providers.iloc[0] / len(df) if len(providers) else 0  # share of the largest provider
years = pd.to_numeric(df["year"], errors="coerce") if "year" in df.columns else pd.Series(dtype=float)
missing = df.isna().mean()                                         # share of empty cells per column
gaps = [c for c in df.columns if missing[c] > 0.5]                 # mostly empty columns

def example(col):
    values = df[col].dropna()
    return str(values.iloc[0])[:40].replace("|", "/") if len(values) else ""

source_line = ("Europeana Search API, searched live on " + today if DATA_SOURCE == "europeana"
               else "Europeana Search API (course copy of the data, obtained 2026-09-15)")
fields = "\n".join(f"| `{c}` | {missing[c]:.0%} | {df[c].nunique()} | {example(c)} |" for c in df.columns)
provider_lines = "\n".join(f"- {name}: {n} records" for name, n in providers.head(10).items())
mark = lambda ok: "x" if ok else " "                               # tick boxes for the checklist

profile = f"""# Dataset profile: {DATASET_PREFIX}

*Created {today} by the course notebook 03_build_dataset. Fill in every line marked "(fill in)".*

## Source and provenance
- **Source:** {source_line}
- **Query:** the creator name forms {", ".join(repr(v) for v in CREATOR_VARIANTS)}
- **Who provided the records** (top 10):
{provider_lines}

## Scope
- **Records:** {len(df)}
- **Types:** {", ".join(f"{k} ({v})" for k, v in df["type"].value_counts().items()) if "type" in df.columns else "(fill in)"}
- **Countries:** {df["country"].nunique() if "country" in df.columns else "(fill in)"}
- **Years:** {f"{int(years.min())} to {int(years.max())} ({years.notna().mean():.0%} of records have a year)" if years.notna().any() else "(no years)"}

## Metadata: the fields
| Field | Missing | Distinct values | Example |
|---|---|---|---|
{fields}

## Rights
- Europeana publishes its metadata under CC0. Rights for the items themselves vary from record to record:
 (fill in what you found, for example on the record pages)

## Gaps and limitations
- Mostly empty fields: {", ".join(gaps) if gaps else "none"}
- The largest provider supplies {top_share:.0%} of the records{" (it dominates the dataset)" if top_share > 0.5 else ""}.
- (fill in anything else you noticed)

## Intended use
(fill in: your research question, and which fields you need for it)

## Suitability checklist
- [{mark(len(df) >= 30)}] Enough records for the question (at least about 30)
- [{mark(top_share <= 0.5)}] No single provider supplies more than half of the records
- [ ] The fields my question needs are present and mostly filled in (fill in)
- [ ] The time, place, and type coverage fits my question (fill in)
- [ ] The licence allows my intended use (fill in)
- [ ] I know what is missing and can state it as a limitation (fill in)
"""

Path("outputs").mkdir(exist_ok=True)
profile_path = Path("outputs") / "dataset_profile.md"
profile_path.write_text(profile, encoding="utf-8")
print("Saved:", profile_path)
print(profile[:900] + "\n...")
```
