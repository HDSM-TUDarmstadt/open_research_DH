### Part 1 · The TIER protocol in brief

[TIER Protocol 4.0](https://www.projecttier.org/tier-protocol/protocol-4-0/) is a documentation standard.
A reproducible project, it says, has:

- a **clear folder structure**;
- clear **documentation files** (a README, metadata, a data appendix that describes the data);
- clear **command files**: the scripts or notebooks that do all the work, in order;
- everything gathered into one **reproducible compendium**, a single package someone else can run.

A typical layout looks like this. The project you set up in Module 2 already follows it:

| TIER component | In your project |
|---|---|
| Original data | `data/raw` |
| Processed (analysis) data | `data/processed` |
| Command files | `notebooks` |
| Documentation | `README.md` and `docs` |
| Results | `outputs` |

### Part 3 · Fix the gaps

Almost everyone finds gaps, and that is the point of review. Typical fixes:

- a notebook without explanation: add a first text cell that says what it does;
- a processing step written down nowhere: add it to the workflow in the README;
- no data description: fill in `docs/data_description.md`;
- data too large for GitHub: explain in the README where it comes from and how to get it;
- no environment file: create `requirements.txt` listing the libraries you use (for this course:
 `pandas`, `matplotlib`, `requests`).

Then **run the check again**. The notebook compares the new results with your last report and adds every
change (for example *Licence: MISSING > OK*) to `docs/revision_log.md`: your **revision log**. Add a few
words on what you did. Anything you cannot fix goes into `docs/unresolved_limits.md`, with the reason and
what it affects; a reader needs to know. Then commit and push.

### Part 5 · The peer test

Find a partner and swap repositories. Your partner tries to run your project **while you explain nothing**:
the project has to speak for itself. They record what happens in the
[reproducibility test template](reproducibility-test-template.md): for each step the expected result, what
actually happened, and a status (works, partly, fails), plus the biggest documentation gap, the biggest
technical problem, and one thing done well. Then you fix what they found.

### 2 · Run the checks

Each check adds one line to the report. Read the comments to see what each one looks for.

```python
import json                                          # notebooks are stored as JSON text
import re                                            # regular expressions, to search text for patterns

report = []                                          # the report: one (status, item, detail) per check

def note(status, item, detail=""):
    """Add one line to the report."""
    report.append((status, item, detail))

def first_existing(*names):
    """Return the first of these folders that exists inside the project, or None."""
    for name in names:
        if (PROJECT_FOLDER / name).is_dir():
            return PROJECT_FOLDER / name
    return None

def real_files(folder):
    """Files in a folder, ignoring the small README files that only describe the folder."""
    return [f for f in folder.rglob("*") if f.is_file() and f.name != "README.md" and ".git" not in f.parts]

# --- 1. README: does it explain the project? ---------------------------------------------
readme = PROJECT_FOLDER / "README.md"
if not readme.exists():
    note("MISSING", "README", "Add a README.md that says what the project is and how to run it.")
else:
    text = readme.read_text(encoding="utf-8", errors="ignore").lower()
    topics = {"research question": ["question"], "data source": ["source", "data"],
              "how to run / workflow": ["run", "workflow", "notebook"], "licence": ["licen"]}
    absent = [t for t, words in topics.items() if not any(w in text for w in words)]
    template_left = [phrase for phrase in [
        "research question and scope: what you investigate",          # instructions from the Module 2 template
        "the steps from the original data to the result, one line each"] if phrase in text]
    if template_left:
        note("CHECK", "README", "Still contains the template's instructions: replace them with your own text.")
    elif len(text) < 800 or absent:
        note("CHECK", "README", "Short or incomplete. Missing topics: " + (", ".join(absent) or "none, but add more detail"))
    else:
        note("OK", "README", "Explains the question, the data, how to run it, and the licence.")

# --- 2. Original (raw) data ----------------------------------------------------------------
raw = first_existing("data/raw", "data/original", "original_data", "original data", "raw")
if raw is None:
    note("MISSING", "Original data folder", "Keep the data exactly as obtained in data/raw.")
elif not real_files(raw):
    note("CHECK", "Original data folder", f"{raw.name} is empty. If the data is too large, say in the README how to obtain it.")
else:
    note("OK", "Original data folder", f"{len(real_files(raw))} file(s) in {raw.relative_to(PROJECT_FOLDER)}")

# --- 3. Processed data ---------------------------------------------------------------------
processed = first_existing("data/processed", "processed", "analysis_data", "analysis data")
if processed is None:
    note("MISSING", "Processed data folder", "Save data produced by the notebooks in data/processed.")
else:
    note("OK", "Processed data folder", str(processed.relative_to(PROJECT_FOLDER)))

# --- 4. Notebooks (the command files) --------------------------------------------------------
notebooks = sorted(p for p in PROJECT_FOLDER.rglob("*.ipynb")
                   if ".ipynb_checkpoints" not in p.parts and ".git" not in p.parts)
if not notebooks:
    note("MISSING", "Notebooks", "No notebooks found. Put them in the notebooks folder.")
else:
    unnumbered = [p.name for p in notebooks if not p.name[:2].isdigit()]
    if unnumbered:
        note("CHECK", "Notebook names", "Number them in the order to run them, e.g. 01_obtain.ipynb: " + ", ".join(unnumbered))
    else:
        note("OK", "Notebook names", f"{len(notebooks)} notebook(s), numbered in run order.")

# --- 5-7. Inside each notebook ------------------------------------------------------------------
ABSOLUTE = re.compile(r"(?<![A-Za-z])[A-Za-z]:[\\/]{1,2}[\w .-]+[\\/]|/(Users|home)/[A-Za-z]")   # a Windows drive path, or a Mac/Linux home folder
no_intro, absolute, out_of_order = [], [], []
for nb_path in notebooks:
    cells = json.loads(nb_path.read_text(encoding="utf-8", errors="ignore")).get("cells", [])
    if not cells or cells[0].get("cell_type") != "markdown":       # should start with an explanation
        no_intro.append(nb_path.name)
    code = ["".join(c.get("source", "")) for c in cells if c.get("cell_type") == "code"]
    if any(ABSOLUTE.search(src) for src in code):                    # paths that only work on one computer
        absolute.append(nb_path.name)
    counts = [c.get("execution_count") for c in cells
              if c.get("cell_type") == "code" and c.get("execution_count")]
    if counts and counts != sorted(counts):                          # numbers jump around: run out of order
        out_of_order.append(nb_path.name)

if notebooks:
    note("CHECK" if no_intro else "OK", "Notebook explanations",
         ("Start these with a text cell saying what they do: " + ", ".join(no_intro)) if no_intro
         else "Every notebook starts with an explanation.")
    note("FIX" if absolute else "OK", "File paths",
         ("Paths that only work on your computer in: " + ", ".join(absolute) + ". Use paths relative to the project folder.")
         if absolute else "No computer-specific paths found.")
    note("CHECK" if out_of_order else "OK", "Run order",
         ("Cells were run out of order in: " + ", ".join(out_of_order) + ". Use Kernel > Restart & Run All.")
         if out_of_order else "No sign of cells run out of order.")

# --- 8. Environment file ------------------------------------------------------------------------
env = [n for n in ["requirements.txt", "environment.yml", "pyproject.toml"] if (PROJECT_FOLDER / n).exists()]
note("OK" if env else "MISSING", "Environment file",
     ", ".join(env) if env else "Add requirements.txt listing the libraries you use (e.g. pandas, matplotlib, requests).")

# --- 9. Documentation ---------------------------------------------------------------------------
docs = PROJECT_FOLDER / "docs"
doc_files = real_files(docs) if docs.is_dir() else []
note("OK" if doc_files else "MISSING", "Documentation",
     f"{len(doc_files)} file(s) in docs" if doc_files else "Add a docs folder with a data description and a decision log.")

# --- 10. Licence --------------------------------------------------------------------------------
licence = [p.name for p in PROJECT_FOLDER.iterdir() if p.is_file() and p.name.upper().startswith(("LICENSE", "LICENCE"))]
note("OK" if licence else "MISSING", "Licence", ", ".join(licence) if licence else "Add a LICENSE file so others know how they may reuse your work.")

# --- 11. Large files ----------------------------------------------------------------------------
big = [(p, p.stat().st_size / 1_000_000) for p in PROJECT_FOLDER.rglob("*")
       if p.is_file() and ".git" not in p.parts and p.stat().st_size > LARGE_FILE_MB * 1_000_000]
too_big = [f"{p.name} ({mb:.0f} MB)" for p, mb in big if mb > 100]
large = [f"{p.name} ({mb:.0f} MB)" for p, mb in big if mb <= 100]
if too_big:
    note("FIX", "File sizes", "GitHub refuses files over 100 MB: " + ", ".join(too_big))
elif large:
    note("CHECK", "File sizes", f"Over {LARGE_FILE_MB} MB (too big for the GitHub website): " + ", ".join(large))
else:
    note("OK", "File sizes", f"No file over {LARGE_FILE_MB} MB.")

print("Checks done:", len(report))
```

### 3 · Read the report

```python
for status, item, detail in report:                  # print one line per check
    print(f"[{status}]".ljust(10), item.ljust(24), detail)

summary = {s: sum(1 for r in report if r[0] == s) for s in ["OK", "CHECK", "MISSING", "FIX"]}
print("\nSummary:", ", ".join(f"{k}: {v}" for k, v in summary.items()))
```

### 4 · Save the report, the revision log, and your unresolved limits

The report goes into your project's `docs` folder. When you run this notebook again after fixing things,
it compares the new results with the last report and adds the differences to `docs/revision_log.md`:
your **revision log**. It also creates `docs/unresolved_limits.md` once, for the problems you cannot fix:
write them down there, so that a reader knows.

```python
import datetime

today = datetime.date.today().isoformat()
if SAVE_REPORT:
    docs = PROJECT_FOLDER / "docs"
    docs.mkdir(exist_ok=True)
    report_path = docs / "reproducibility_report.md"

    # 1) Compare with the previous report, if there is one: that is your revision log.
    previous = {}
    if report_path.exists():
        for line in report_path.read_text(encoding="utf-8").splitlines():
            parts = [p.strip() for p in line.strip("|").split("|")]
            if len(parts) == 3 and parts[0] in ("OK", "CHECK", "MISSING", "FIX"):
                previous[parts[1]] = parts[0]
    changes = [(item, previous[item], status) for status, item, _ in report
               if item in previous and previous[item] != status]
    if previous:
        print("Changes since your last check:" if changes else "No changes since your last check.")
        for item, before, after in changes:
            print(f"   {item}: {before} -> {after}")
        log = docs / "revision_log.md"
        if not log.exists():
            log.write_text("# Revision log\n\nWhat changed between reproducibility checks.\n\n"
                           "| Date | Item | Before | After | What I did |\n|---|---|---|---|---|\n", encoding="utf-8")
        with log.open("a", encoding="utf-8") as f:
            for item, before, after in changes:
                f.write(f"| {today} | {item} | {before} | {after} | (fill in) |\n")
        print("Updated:", log)

    # 2) Write the new report.
    lines = ["# Reproducibility report", "",
             f"Checked on {today} with the course notebook 07_reproducibility_check.", "",
             "| Status | Item | Detail |", "|---|---|---|"]
    lines += [f"| {s} | {i} | {d} |" for s, i, d in report]
    report_path.write_text("\n".join(lines) + "\n", encoding="utf-8")
    print("Saved:", report_path)

    # 3) A place for the limits you cannot fix (created once, never overwritten).
    limits = docs / "unresolved_limits.md"
    if not limits.exists():
        limits.write_text("# Unresolved limits\n\nWhat you could not fix, and why. A reader needs to know.\n\n"
                          "| Limit | Why it cannot be fixed now | What it affects |\n|---|---|---|\n| | | |\n",
                          encoding="utf-8")
        print("Created:", limits)
else:
    print("Demonstration run: no report saved.")
```
