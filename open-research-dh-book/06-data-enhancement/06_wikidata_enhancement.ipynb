```python
from pathlib import Path            # cross-platform file paths
import pandas as pd                   # tables
import requests                       # HTTP requests to the Wikidata API
import time                           # to pause politely between requests

# Filenames derived from the prefix.
FILTERED_FILE = f"{DATASET_PREFIX}_filtered.csv"          # preferred input (Module 4 output)
ENHANCED_FILE = f"{DATASET_PREFIX}_enhanced.csv"          # output table
MAP_FILE      = f"{ENTITY_COLUMN}_wikidata_map.csv"       # reusable entity -> Wikidata lookup

def find_dataset(name):
    """Return the path to `name`, looking in ./datasets and ../datasets."""
    for base in [Path("datasets"), Path("../datasets")]:
        if (base / name).exists():
            return base / name
    raise FileNotFoundError(name)

def datasets_dir():
    """Return the datasets/ folder to save into."""
    for base in [Path("datasets"), Path("../datasets")]:
        if base.exists():
            return base
    Path("datasets").mkdir(parents=True, exist_ok=True); return Path("datasets")

def load_input():
    """Load the dataset to enhance (filtered first, then earlier files as a fallback)."""
    for name in [FILTERED_FILE, "van_gogh_filtered.csv", "van_gogh_combined.csv"]:
        try:
            path = find_dataset(name); print("Loaded", path)
            return pd.read_csv(path)
        except FileNotFoundError:
            continue
    raise FileNotFoundError("No dataset found, run Modules 3 and 4 first.")

df = load_input()

# Safety check: the column to reconcile must exist.
if ENTITY_COLUMN not in df.columns:
    raise KeyError(f"Column {ENTITY_COLUMN!r} not found. Available columns: {list(df.columns)}")

# The unique, non-empty values to reconcile (reconcile each once, not once per row).
entities = sorted(df[ENTITY_COLUMN].dropna().unique())
if SAMPLE_ONLY:
    entities = entities[:3]           # a quick test with just the first three
print(f"{len(entities)} unique {ENTITY_COLUMN!r} values to reconcile.")
entities
```

### 2 · Match by hand first (the skill before the code)

Before automating, do it once by hand so you understand the judgement involved.

1. Go to <https://www.wikidata.org> and search one value from the list above.
2. Often several items match. **Disambiguate** using the item's **label**, **description**, and
 **statements** (type, location) plus what you already know from your data.
3. Decide: **matched** (confident), **uncertain** (plausible but unsure), or **not found**.

> **Prompt you could use (then check it):** *Given a name, search Wikidata, return the best-matching
> item's Q-ID, label and description, and flag if it does not look like something that would be expected.*

The functions below do this and record a **match_status**, being honest about uncertainty is part
of the scholarship.

```python
WD_API = "https://www.wikidata.org/w/api.php"          # the Wikidata API endpoint

# Wikidata asks every client to identify itself; a request without a User-Agent is often
# rejected with HTTP 403. Put your own project name / contact here if you like.
HEADERS = {"User-Agent": "OpenResearchDH-course/1.0 (teaching example)"}

# Words that suggest a "matched" item for the DEFAULT (institution) case. If you reconcile
# people or places instead, change these hints to fit (or set them to () to skip the check).
MATCH_HINTS = ("museum", "library", "gallery", "university", "archive",
               "collection", "institute", "foundation")

def wd_get(params, tries=5):
    """GET from the Wikidata API, retrying politely on HTTP 429 (rate limit) with backoff."""
    for attempt in range(tries):                          # try up to `tries` times
        r = requests.get(WD_API, headers=HEADERS, params=params, timeout=30)
        if r.status_code == 429:                          # 429 = too many requests
            wait = int(r.headers.get("Retry-After", 2 ** attempt))   # server may tell us how long
            print(f"    rate-limited, waiting {min(wait, 30)}s...")
            time.sleep(min(wait, 30))                     # wait (capped at 30s), then retry
            continue
        r.raise_for_status()                              # any other error: stop clearly
        return r.json()                                   # success: return parsed JSON
    r.raise_for_status()                                  # retries exhausted: raise the 429
    return r.json()

def wd_search(term, limit=5):
    """Search Wikidata for `term`; return candidate items as dicts {id, label, description}."""
    data = wd_get({"action": "wbsearchentities", "search": term, "language": LANG,
                   "uselang": LANG, "format": "json", "limit": limit})
    return data.get("search", [])                         # the list of candidates (may be empty)

def wd_feature(qid, prop=FEATURE_PROPERTY):
    """Return the value of property `prop` on item `qid` as readable text, or None.

    Works for the common kinds of value: another item (such as a country), a date (such as inception),
    coordinates, a number, or text.
    """
    data = wd_get({"action": "wbgetentities", "ids": qid, "props": "claims", "format": "json"})
    claims = data["entities"][qid].get("claims", {})        # all statements on the item
    values = claims.get(prop)                               # the ones for our property (e.g. P17)
    if not values or "datavalue" not in values[0]["mainsnak"]:   # no statement, or no known value
        return None
    value = values[0]["mainsnak"]["datavalue"]["value"]     # the value of the first statement
    if isinstance(value, dict) and "id" in value:           # another item, such as a country
        target = value["id"]
        data2 = wd_get({"action": "wbgetentities", "ids": target, "props": "labels",
                        "languages": LANG, "format": "json"})
        return data2["entities"][target]["labels"].get(LANG, {}).get("value", target)   # its name
    if isinstance(value, dict) and "time" in value:         # a date, stored like "+1792-08-10T00:00:00Z"
        t = value["time"]
        return t[1:5] if t.startswith("+") else t[:5]      # keep the year (with a minus sign if BCE)
    if isinstance(value, dict) and "latitude" in value:     # coordinates
        return f'{value["latitude"]:.5f}, {value["longitude"]:.5f}'
    if isinstance(value, dict) and "amount" in value:       # a number
        return value["amount"].lstrip("+")
    if isinstance(value, dict) and "text" in value:         # text in a particular language
        return value["text"]
    return str(value)                                       # plain text, such as an identifier

import datetime
TODAY = datetime.date.today().isoformat()

def describe(qid):
    """Label and description of one Wikidata item (used for matches decided by hand)."""
    data = wd_get({"action": "wbgetentities", "ids": qid, "props": "labels|descriptions",
                   "languages": LANG, "format": "json"})
    item = data["entities"][qid]
    return (item.get("labels", {}).get(LANG, {}).get("value"),
            item.get("descriptions", {}).get(LANG, {}).get("value"))

def reconcile_one(term):
    """Reconcile one value. Returns the Q-ID and labels, the feature, a match status, the evidence for the
    decision, the candidates that were rejected, the source link, and any error."""
    result = {"qid": None, "wikidata_label": None, "wikidata_description": None, FEATURE_NAME: None,
              "match_status": None, "evidence": None, "rejected_candidates": "", "source": None,
              "retrieved": TODAY, "error": None}
    try:
        if term in MANUAL_MATCHES:                           # you decided this one by hand
            qid = MANUAL_MATCHES[term]
            label, desc = describe(qid)
            hits = [{"id": qid, "label": label, "description": desc}]
            status, evidence = "matched by hand", "checked by hand on Wikidata"
        else:
            hits = wd_search(term)                           # search Wikidata
            status = evidence = None
    except requests.exceptions.ConnectionError:              # no internet
        result.update(match_status="offline", error="no connection to Wikidata"); return result
    except requests.exceptions.Timeout:                      # server too slow
        result.update(match_status="error", error="request timed out"); return result
    except Exception as e:                                   # anything else: report the error as it is
        result.update(match_status="error", error=f"{type(e).__name__}: {e}"); return result

    if not hits:                                             # search returned nothing
        result.update(match_status="not_found", evidence="no Wikidata item found for this name")
        return result

    top, others = hits[0], hits[1:]                          # the best candidate, and the rest
    desc = (top.get("description") or "").lower()
    if status is None:                                       # decided by the rule, not by hand
        found = [h for h in MATCH_HINTS if h in desc]        # institution words in the description
        looks_right = (not MATCH_HINTS) or bool(found)
        status = "matched" if looks_right else "uncertain"
        evidence = (f"best search result; its description contains '{found[0]}'" if found
                    else "best search result, but its description does not look like an institution")
    result.update(qid=top["id"], wikidata_label=top.get("label"), wikidata_description=top.get("description"),
                  match_status=status, evidence=evidence,
                  rejected_candidates="; ".join(f"{h['id']} ({h.get('label')}: {h.get('description') or 'no description'})"
                                                for h in others),
                  source=f"https://www.wikidata.org/wiki/{top['id']}")
    try:
        result[FEATURE_NAME] = wd_feature(top["id"])         # pull the requested feature
    except Exception:
        result[FEATURE_NAME] = None                          # feature lookup failed: leave it blank
    return result
```

### 3 · Reconcile the unique entities

We reconcile each **unique** value once, building a reusable mapping table. Inspect it before merging:
check the labels/descriptions look right, and review anything marked *uncertain*.

```python
records = []                                               # one dict per unique value
for name in entities:                                      # go through each unique value
    info = reconcile_one(name)                             # reconcile it
    info[ENTITY_COLUMN] = name                             # remember which value this was
    records.append(info)                                   # keep the result
    reason = f'  ({info["error"]})' if info.get("error") else ""   # show the error if any
    print(f'  {name!r:45s} -> {info["match_status"]:9s} '
          f'{info.get("qid") or ""} {info.get(FEATURE_NAME) or ""}{reason}')
    time.sleep(1)                                          # pause ~1s: be polite to the free API

# Build the mapping table with a tidy column order.
mapping = pd.DataFrame(records)[[ENTITY_COLUMN, "qid", "wikidata_label", "wikidata_description",
                                 FEATURE_NAME, "match_status", "evidence", "rejected_candidates",
                                 "source", "retrieved"]]
print("\nStatus counts:\n", mapping["match_status"].value_counts())   # how many matched / uncertain / ...

errors = [r["error"] for r in records if r.get("error")]  # collect any error messages
if errors:
    print("\nFirst error reported:", errors[0])           # show the first, to help debugging
mapping
```

### 4 · Save the mapping, merge it back, and write the enhancement table

We save the reusable **entity > Wikidata** lookup table, join it onto every row of the dataset, and write
the **enhancement table**: the result of Module 6. For each entity it records the Q-ID, the status
(how certain the match is), the evidence for the decision, the source, and the candidates that were
rejected.

```python
# Only save if Wikidata was actually reached. If every lookup failed (for example, no internet),
# saving would replace a good earlier result with empty columns, so we stop here instead.
reached = ~mapping["match_status"].isin(["offline", "error"])    # True where Wikidata answered
CORE = [ENTITY_COLUMN, "qid", "wikidata_label", "wikidata_description", FEATURE_NAME, "match_status"]

if not reached.any():
    print("Nothing was saved: Wikidata could not be reached for any entity.")
    print("Connect to the internet and run the notebook again. Your earlier files are unchanged.")
else:
    mapping.to_csv(datasets_dir() / MAP_FILE, index=False)     # the full lookup table, with the evidence

    # Merge the core columns onto every row of the dataset. how="left" keeps every original row.
    enhanced = df.merge(mapping[CORE].rename(columns={FEATURE_NAME: f"wikidata_{FEATURE_NAME}"}),
                        on=ENTITY_COLUMN, how="left")
    enhanced.to_csv(datasets_dir() / ENHANCED_FILE, index=False)   # save the enhanced dataset
    print("Saved:", MAP_FILE, "and", ENHANCED_FILE)
    if not reached.all():
        print(f"Note: {(~reached).sum()} value(s) could not be looked up; run the notebook again later to fill them in.")

    # The enhancement table: the result of Module 6
    clean = lambda v: "" if pd.isna(v) else v                        # show empty cells as blank, not "nan"
    rows = "\n".join(
        f"| {r[ENTITY_COLUMN]} | {clean(r['qid'])} | {r['match_status']} | {clean(r['evidence'])} | "
        f"{clean(r['source'])} | {clean(r['rejected_candidates'])} | {clean(r[FEATURE_NAME])} |"
        for _, r in mapping.iterrows())
    table = f"""# Enhancement table: {ENTITY_COLUMN} linked to Wikidata

*Created {TODAY} by the course notebook 06_wikidata_enhancement. Source: Wikidata (CC0), retrieved {TODAY}.
Feature added: {FEATURE_NAME} (property {FEATURE_PROPERTY}).*

Status counts: {", ".join(f"{k}: {v}" for k, v in mapping["match_status"].value_counts().items())}.

| {ENTITY_COLUMN} | Q-ID | Status (certainty) | Evidence for the decision | Source | Rejected candidates | {FEATURE_NAME} |
|---|---|---|---|---|---|---|
{rows}
"""
    Path("outputs").mkdir(exist_ok=True)
    table_path = Path("outputs") / "enhancement_table.md"
    table_path.write_text(table, encoding="utf-8")
    print("Saved:", table_path)
    display(enhanced.head())
```
