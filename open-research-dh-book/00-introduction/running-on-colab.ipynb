# Running the Notebooks on Google Colab

Google Colab is a free, browser-based way to run the course notebooks with **nothing to install**. It works
on any computer with a web browser, including older computers that cannot install Anaconda.

## Steps

1. **Open Colab.** Go to <https://colab.research.google.com> and sign in with a Google account.
2. **Open a notebook.** Use one of the direct links below. Or, in Colab, click *File > Open notebook*,
 choose the **GitHub** tab, paste `https://github.com/HDSM-TUDarmstadt/open_research_DH`, press **Enter**, and
 click the notebook you want.
3. **Run the first code cell, *Google Colab setup*.** It downloads the course into Colab so the notebook can
 find the `datasets` folder. If Colab warns that the notebook was not written by Google, click
 **Run anyway**.
4. **Run the rest** from top to bottom: press **Shift + Enter** on each cell (or *Runtime > Run all*).

## Saving your work

Colab resets everything when the session ends (re-running the setup cell brings the course back).

- **To keep a notebook with your changes:** *File > Save a copy in Drive*. The copy is in your own Google
 Drive, not in the course.
- **To keep files the notebook created** (CSV files, charts): open the **Files** panel on the left
 (the folder icon), right-click the file, and choose **Download**.

## Optional: keep everything in Google Drive
