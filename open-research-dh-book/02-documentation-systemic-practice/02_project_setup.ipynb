# Module 2 · Documenting as a systemic practice: the research workflow

Documentation is not paperwork you do at the end: it is research you do as you go. This module shows what
to document and why, how to turn an interest into a data-driven research question, and how to keep a
documented project under version control on GitHub.

| **At a glance** | |
|---|---|
| **Aim** | Documentation as a core research practice that supports transparency, reproducibility, and long-term reuse in open humanities research. |
| **Module activity** | Map the project from question and source selection to output; begin a README and decision log; record changes using your chosen route. |
| **Time needed** | about 2 to 3 hours |
| **You will need** | your setup from Module 0 (Path A is best for this module); a free GitHub account and GitHub Desktop, which you set up in Step 3 |

**How this module works.** Go through Steps 1 to 6 in order; each step says exactly what to do. Everything is on this page: the text, the videos, and (in Step 4) the code you run. The learning outcomes are the checklist in Step 5, and further reading is at the end.

Two words first. A **commit** is a saved checkpoint of your project, with a short message that says what
you changed. A **push** sends your commits from your computer to GitHub. GitHub Desktop is a program that
does both with buttons, so you do not need to type commands.

**Two routes for keeping track of changes.** The **GitHub route** (Parts A to G) is recommended: every
commit is kept in your project's history. If you cannot use GitHub for any reason, take the **change-log route**: run the notebook in
Part E with `CREATE_IF_MISSING = True` (on Colab this happens by itself), and record every change in

### Part G · Commit and push

**One rule to remember:** GitHub is for code and documentation, not for large raw data. The website
accepts files up to 25 MB, and GitHub refuses files over 100 MB. Keep big raw files on your computer and/or on an online drive where others can
access them, and describe in the README where they come from and how to get them.

### 2 · Create the folders

Each folder gets a short README that says what belongs in it. This matters because GitHub does not keep
empty folders. If a folder already exists, nothing in it is changed.

```python
FOLDERS = {                                        # folder name: what belongs in it
    "data/raw":       "Original data, exactly as obtained. Never edit these files.",
    "data/processed": "Data produced by the notebooks: cleaned, filtered, or enriched.",
    "notebooks":      "Jupyter notebooks, numbered in the order they should be run (01_..., 02_...).",
    "docs":           "Documentation: data description, decision log, AI-use record, reports.",
    "outputs":        "Charts, tables, and other results.",
}

for folder, purpose in FOLDERS.items():            # go through the folders one by one
    folder_path = PROJECT_FOLDER / folder          # the full path of this folder
    folder_path.mkdir(parents=True, exist_ok=True) # create it (no error if it already exists)
    readme = folder_path / "README.md"             # the small README inside the folder
    if not readme.exists():                        # only write it if it is not there yet
        readme.write_text(f"# {folder}\n\n{purpose}\n", encoding="utf-8")
    print("[OK]", folder)
```

### 3 · Write the project README

The README is the first thing a reader sees. Its sections follow the video *Open Research and
Documentation*: what the project is, where the data comes from, what it contains, how it was made, the
workflow, the decisions, and the results and limitations.

If your repository only has the short README that GitHub creates, the sections are added below it. If you
have already written your own README, it is left as it is.

```python
README_SECTIONS = f"""## What is this project?
Research question and scope: what you investigate, and which part of the collection or dataset you use.

## Where does the data come from?
- Source (institution or platform, with a link):
- Who created or provided it:
- Date accessed:
- Query, version, or download method:
- Licence and conditions of use:

## What does the data contain?
What one record represents, the main fields and what they mean, and what is missing or uncertain.
Details: [docs/data_description.md](docs/data_description.md).

## How was it made? (methods and tools)
The software, notebooks, libraries, and services used, with versions if they matter.
To install the libraries: `pip install -r requirements.txt`.

## Workflow
The steps from the original data to the result, one line each, with the notebook that does each step:
1. Obtain the data: `notebooks/...`
2. ...

## Decisions
Important choices, and why they were made: [docs/decision_log.md](docs/decision_log.md).
Use of AI: [docs/ai_use_record.md](docs/ai_use_record.md).

## Results, limitations, and possible uses

## Folder structure
- `data/raw`: original data, never edited
- `data/processed`: data produced by the notebooks
- `notebooks`: the notebooks, in the order to run them
- `docs`: documentation
- `outputs`: charts and tables

## Licence
"""

readme = PROJECT_FOLDER / "README.md"              # the main README of the project
header = f"{YOUR_NAME} · started {TODAY}"          # a line with your name and today's date

if not readme.exists():                            # no README at all: write a complete one
    readme.write_text(f"# {PROJECT_TITLE}\n\n{header}\n\n{README_SECTIONS}", encoding="utf-8")
    print("[CREATED] README.md")
else:
    old = readme.read_text(encoding="utf-8")       # what is in the README now
    if len(old) < 400 and "## What is this project?" not in old:   # only GitHub's short starter text
        readme.write_text(old.rstrip() + f"\n\n{header}\n\n{README_SECTIONS}", encoding="utf-8")
        print("[UPDATED] README.md (sections added below GitHub's short README)")
    else:                                          # you already wrote it: do not touch it
        print("[KEPT]    README.md (already written, not changed)")
```

### 4 · Write the documentation templates

Five templates go into `docs`: a **data description**, a **decision log**, a **record for AI use**, a
**workflow map** (from question to output, step by step), and a **change log** (for the route without
GitHub). Two small helper files go into the main folder: `requirements.txt` (which libraries the project
needs) and `.gitignore` (which files Git should ignore). Files that already exist are kept unchanged.

```python
def write_new(relative_path, text):
    """Create a file in the project, but never overwrite a file that already exists."""
    file_path = PROJECT_FOLDER / relative_path     # the full path of the file
    if file_path.exists():                         # it is already there
        print("[KEPT]   ", relative_path)
    else:
        file_path.parent.mkdir(parents=True, exist_ok=True)   # make sure its folder exists
        file_path.write_text(text, encoding="utf-8")          # write the text into it
        print("[CREATED]", relative_path)


write_new("docs/data_description.md", """# Data description

## Source
- Institution or platform (with link):
- Date accessed:
- Query, version, or download method:
- Licence and conditions of use:

## What one record represents

## Fields
| Field | What it means | Example | Missing or uncertain? |
|---|---|---|---|
| | | | |

## Known gaps and limitations
""")

write_new("docs/decision_log.md", f"""# Decision log

One entry per decision: what you decided, why, and what it affects. Add new entries at the bottom.

| Date | Decision | Why | What it affects |
|---|---|---|---|
| {TODAY} | Set up the project structure | To keep raw data, processed data, code, and documentation apart from the start | The whole project |
""")

write_new("docs/ai_use_record.md", """# AI-use record

Copy the block below each time AI shaped your code, data, or text in a way that matters.

## Date · short title
- **Tool and model (with version, if shown):**
- **Purpose:** what you wanted to achieve
- **Role:** a tool that supports your work, or a component that shapes the data?
- **Prompt:** (paste it)
- **What the output does:**
- **What I checked:** (for example row counts before and after, a sample of the result)
- **What I changed:**
- **Limitations and risks I watched for:** (bias, hallucination, privacy, copyright, reproducibility)
""")

write_new("docs/workflow_map.md", f"""# Workflow map

From the research question to the outputs, one row per step. Someone who has never seen the project
should be able to name the inputs and the main steps from this table alone.

**Research question:**
**Main source:**

| Step | Input | Done with (notebook or tool) | Output | Decision recorded in |
|---|---|---|---|---|
| 1. Obtain the data | | | data/raw/... | docs/decision_log.md |
| 2. Select and filter | data/raw/... | | data/processed/... | docs/decision_log.md |
| 3. Analyse and visualise | data/processed/... | | outputs/... | |
| 4. Enhance (optional) | | | | |
| 5. Publish | | | | |
""")

write_new("docs/change_log.md", f"""# Change log

If you use Git and GitHub, your commit history is your record of changes and this file is optional.
If you do not (for example because you work only in Google Colab), record every change here instead.

| Date | What changed | Which files | Why |
|---|---|---|---|
| {TODAY} | Set up the project structure | all | start of the project |
""")

write_new("requirements.txt", "pandas\nmatplotlib\nrequests\n")
write_new(".gitignore", ".ipynb_checkpoints/\n.DS_Store\n__pycache__/\n")
```
