# Well-Documented Example Projects

A short, curated set of real open projects worth studying as models of **good documentation and
reproducibility**. Use them for the *exploring workflows* task in this module, and revisit them in
Module 7 when you audit your own project. For each one, look at *how it is documented*, as well as what
it produced.

## What to look for
When you open a project, check it against the same things this course asks of your own work:

- a **clear folder structure** (raw vs processed data, code, documentation);
- a **README** that states the aim, the data, and how to run the project;
- **data provenance**: where the data came from, the query used, the date accessed, and the licence;
- an **environment file** or plain run instructions, so someone else can reproduce it;
- **recorded decisions**: what was included or excluded, and why;
- **stable identifiers / DOIs** for the data and outputs;
- entities reconciled to **authority files** (e.g. Wikidata) where relevant.

## Reproducible notebooks and collections as data
- **GLAM Workbench**: https://glam-workbench.net (code: https://github.com/glam-workbench). The clearest
 model for this course: dozens of documented Jupyter notebooks that turn galleries, libraries, archives,
 and museums collections into data. Each section states its data source, runs in a shareable environment,
 and is archived with a DOI. Look at how one tool documents where its data comes from and how to re-run it.
- **Living with Machines**: https://github.com/Living-with-machines. A large historical-newspapers
 project whose research code is published as separate repositories, each with a README, an environment,
 and (often) a linked dataset and paper. A good example of splitting a big project into documented,
 citable pieces.

## Documenting the method and the decisions
- **The Data-Sitters Club**: https://datasittersclub.github.io/site/ (code:
 https://github.com/datasittersclub). Radically transparent digital-humanities methods: the authors write
 up every choice, tool, and dead end in plain language. The single best model for *documenting as you go*
 and being honest about uncertainty.
- **Programming Historian**: https://programminghistorian.org. Peer-reviewed, openly licensed tutorials
 that ship with their data and code. Look at how a lesson explains each step so a stranger can follow it,
 and at the project's own sustainability and documentation practices.

## Institutions publishing collections as data
- **Library of Congress Labs**: https://labs.loc.gov. Experiments in publishing collections as data, with
 documented data packages, notebooks, and reports (see the *Collections as Data* and *Computing Cultural
 Heritage in the Cloud* work). A model for how a source institution documents a dataset for reuse.
- **Collections as Data (Always Already Computational)**: https://collectionsasdata.github.io. The
 guiding principles and practical checklists (including the "50 Things" list) for making and documenting
 collections as data. Useful as a standard to compare a project against.

## Documented datasets and data papers
- **Journal of Open Humanities Data**: https://openhumanitiesdata.metajnl.com. Peer-reviewed *data
 papers*: short articles whose job is to document a dataset (its creation, structure, and reuse). Read one
 data paper as a template for the documentation your own dataset deserves.

## Linked open data and authority control
- **Six Degrees of Francis Bacon**: http://sixdegreesoffrancisbacon.com. An open, downloadable
 prosopographical network of early-modern people, built on stable identifiers rather than name strings.
 A concrete example of the authority-control idea from Modules 3 and 6, at scale.

## Reproducibility handbooks and templates
- **The Turing Way**: https://book.the-turing-way.org. A community handbook for reproducible, ethical, and
 collaborative research, with checklists you can apply directly to your own repository.
- **TIER Protocol 4.0**: https://www.projecttier.org/tier-protocol/protocol-4-0/. The documentation
 standard you use in Module 7, with sample projects showing the recommended structure in practice.
