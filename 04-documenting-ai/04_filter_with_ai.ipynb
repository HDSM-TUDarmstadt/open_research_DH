### 3.1 · Write the goal in plain words first

Before you ask an AI assistant for code, write down what you want, in a text cell or a note:

- **Which data?** For example: the combined Van Gogh dataset, `van_gogh_combined.csv`.
- **Which one change?** For example: remove the rows from a reproduction index, because my question is
 about the institutions that hold the works.
- **What should come out?** For example: a new file, `van_gogh_filtered.csv`, and the number of rows
 before and after.

The clearer this is, the better the code you get back. Most of it becomes your prompt.

### 3.2 · Turn it into a good prompt

A good prompt for code:

1. states the **conditions**: *Python, in a Jupyter notebook, using pandas*;
2. is **concrete**: the input file, the exact column names, what you want as output, and where to save it;
3. asks for **short code with a comment on each step**;
4. asks the code to **print what it does**, so you can follow it;
5. sets **limits**: *do not change the original file*;
6. asks for a **small sample** of the result, for example ten rows, so you can inspect it. You can draw a random sample.

For example:

> In a Jupyter notebook, using pandas, load `van_gogh_combined.csv`. Create a new table called `filtered`
> that leaves out every row whose `institution` is in a list I give you. Do not change the original table.
> Print how many rows were removed and how many remain, show the remaining institutions with their counts,
> and save the result as `van_gogh_filtered.csv`. Keep the code concise and comment every line.

### 3.3 · Check the code before you trust it

AI can choose the wrong column, drop too much, use a function that does not exist, or change your data
without saying so. So, every time:

1. **Read** the code. Can you say in your own words what each line does? If not, ask the assistant to
 explain the line.
2. **Run it on a small piece first**, for example the first 20 rows.
3. **Compare the numbers**: rows before, rows removed, rows after. Do they add up? Do they match what you
 expected?
4. **Look at a sample** of the result.

If you get an error, copy the **whole** error message back to the assistant and ask it to explain and fix
it. Then check the new version again.

### 1 · Load the dataset from the previous module

The helper functions below find the shared `datasets/` folder wherever you launched the notebook from,
and load the combined file. Every line is commented.

```python
from pathlib import Path            # safe, cross-platform file paths
import pandas as pd                   # tables (DataFrames)

# Filenames are derived from the prefix, so renaming the creator renames the files.
COMBINED_FILE = f"{DATASET_PREFIX}_combined.csv"   # input from Module 3
FILTERED_FILE = f"{DATASET_PREFIX}_filtered.csv"   # output we will write

# Names we accept when searching for the input (yours first, then course samples).
KNOWN_DATASETS = [COMBINED_FILE, "van_gogh_combined.csv", "van_gogh_europeana.csv", FILTERED_FILE]

def find_dataset(name):
    """Return the path to a data file called `name`, looking in ./datasets and ../datasets."""
    for base in [Path("datasets"), Path("../datasets")]:   # both likely locations
        if (base / name).exists():                         # does the file exist there?
            return base / name                             # yes: return the path
    raise FileNotFoundError(f"Could not find {name} in datasets/ or ../datasets/")

def datasets_dir():
    """Return the datasets/ folder to save into (beside any existing data)."""
    for base in [Path("datasets"), Path("../datasets")]:
        if any((base / n).exists() for n in KNOWN_DATASETS):
            return base
    Path("datasets").mkdir(parents=True, exist_ok=True)
    return Path("datasets")

def load_combined():
    """Load the combined dataset produced in Module 3 (try your prefix, then the sample)."""
    for name in [COMBINED_FILE, "van_gogh_combined.csv", "van_gogh_europeana.csv"]:
        try:
            path = find_dataset(name); print("Loaded", path)
            return pd.read_csv(path)                        # read CSV into a DataFrame
        except FileNotFoundError:
            continue                                        # not there: try the next name
    raise FileNotFoundError("Run Module 3 first, or check the datasets/ folder.")

df = load_combined()                 # the combined table
print("Records:", len(df))           # how many rows we loaded

# A safety check: make sure the column we plan to filter on actually exists.
if ENTITY_COLUMN not in df.columns:
    raise KeyError(f"Column {ENTITY_COLUMN!r} not found. Available columns: {list(df.columns)}")

# Show the most common values so you can SEE what might need excluding.
df[ENTITY_COLUMN].value_counts().head(8)
```

### 2 · Decide the filter, and say why

Our question is about **institutions that hold the works**. One value in the data is a **reproduction
index**, not a holding museum, so for *this* question we set it aside. This is a **documented decision**: we keep the original and record what we did.

> **Prompt you could use (then check it!):** *In pandas, from `df`, create `filtered` that removes rows
> whose `institution` is in a list of reproduction/aggregator providers I give you. Do not modify the
> original. Print how many rows were removed and how many remain.*
>
> **Always check AI code.** Read it, run it on a small piece, and confirm the row counts make sense.

#### Your turn: filter out the reproduction index

**Task:** make a new table `my_filtered` without the rows whose institution is in `EXCLUDE_VALUES`, and without changing `df`. You can write it yourself or ask an AI assistant (see Step 3 above for a prompt), but always remember to check the result.

**Hint:** `df[ENTITY_COLUMN].isin(EXCLUDE_VALUES)` is True for the rows to set aside; a `~` in front means *not*. Add `.copy()` at the end.

**You should get:** 34 rows left, from 10 institutions.

Try it in the empty cell below. Then unfold the **solution** cell under it, compare, and run the solution
cell too, because the next steps use it.

```python
# Your attempt: write your code here, then press Shift + Enter
```

```python
#@title Solution (unfold to see the code)
# df[ENTITY_COLUMN].isin(EXCLUDE_VALUES) is True for rows whose value is in our exclude list.
# The ~ in front means "NOT", so we keep the rows that are NOT in the list.
# .copy() makes a separate table, so we never change the original df (good practice).
filtered = df[~df[ENTITY_COLUMN].isin(EXCLUDE_VALUES)].copy()

removed = len(df) - len(filtered)                         # how many rows we set aside
print(f"Removed {removed} excluded rows; {len(filtered)} remain.")
filtered[ENTITY_COLUMN].value_counts()                    # what is left, most common first
```

### 3 · A second, gentler tidy (optional)

Light tidying only: drop rows with no entity value so empty groups do not distort the counts. We never
invent or change values. Turn this off by setting `DROP_MISSING_ENTITY = False` in the configuration.

```python
if DROP_MISSING_ENTITY:                                    # only if you asked for it
    # notna() is True for non-empty cells; the second test drops cells that are just "".
    keep = filtered[ENTITY_COLUMN].notna() & (filtered[ENTITY_COLUMN].astype(str).str.strip() != "")
    filtered = filtered[keep].copy()                      # keep only those rows

print("After tidying:", len(filtered), "records across",
      filtered[ENTITY_COLUMN].nunique(), "distinct", ENTITY_COLUMN + "s")
```

### 4 · Save the filtered dataset for Module 5

```python
out_path = datasets_dir() / FILTERED_FILE                 # e.g. datasets/van_gogh_filtered.csv
filtered.to_csv(out_path, index=False)                    # write it (no row-number column)
print("Saved ->", out_path)
```

### 5 · AI-use statement and verification log

This is the result of Module 4. The cell below checks the filtering automatically (rows before and after,
nothing excluded left behind, original unchanged) and writes a **verification log** together with an
**AI-use statement** to `outputs/ai_use_and_verification.md`. Open the file and complete the lines marked
*(fill in)*: which tool and model you used, your prompt, and what you changed.

```python
import datetime

original_again = load_combined()                                  # read the original file once more
checks = [
    ("Rows in the original data", len(df), ""),
    ("Rows set aside (EXCLUDE_VALUES)", int(df[ENTITY_COLUMN].isin(EXCLUDE_VALUES).sum()), ""),
    ("Rows dropped because the entity was empty", int(len(df) - df[ENTITY_COLUMN].isin(EXCLUDE_VALUES).sum() - len(filtered)), ""),
    ("Rows in the filtered data", len(filtered), ""),
    ("Excluded values left in the filtered data", int(filtered[ENTITY_COLUMN].isin(EXCLUDE_VALUES).sum()), "should be 0"),
    ("Original file unchanged (same number of rows)", "yes" if len(original_again) == len(df) else "NO", "should be yes"),
]
rows_line = "\n".join(f"| {what} | {value} | {note} |" for what, value, note in checks)

statement = f"""# AI-use statement and verification log: filtering

*Created {datetime.date.today().isoformat()} by the course notebook 04_filter_with_ai.*

## AI-use statement
- **Task:** filter the dataset so that it fits the research question.
- **Role of AI:** a tool. It drafted the filtering code; the decision what to filter was mine.
- **Tool and model (with version, if shown):** (fill in)
- **Prompt:** (fill in)
- **What I changed in the code after checking it:** (fill in)
- **Limitations and risks I watched for:** (fill in)

## Verification log
| Check | Result | Note |
|---|---|---|
{rows_line}

Set aside, not deleted: {", ".join(EXCLUDE_VALUES)}. The original file is kept unchanged.
"""
Path("outputs").mkdir(exist_ok=True)
statement_path = Path("outputs") / "ai_use_and_verification.md"
statement_path.write_text(statement, encoding="utf-8")
for what, value, note in checks:
    print(f"{what:48s} {value}")
print("\nSaved:", statement_path)
```
