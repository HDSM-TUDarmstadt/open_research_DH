### Configuration

The only cell you might change. To look at your own dataset, put its file name in `DATASET_FILE` and the
names of its columns in the other two lines.

```python
DATASET_FILE    = "van_gogh_combined.csv"   # the dataset to look at (in the datasets folder)
PROVIDER_COLUMN = "institution"             # the column that names who provided each record
COUNTRY_COLUMN  = "country"                 # the column that names the provider's country

print("Looking at", DATASET_FILE)
```

### 1 · Open the dataset

```python
from pathlib import Path        # a safe way to work with file paths
import pandas as pd             # the library for tables
import matplotlib.pyplot as plt # the library for charts

def find_dataset(name):
    """Return the path to a file in the course's datasets folder, wherever the notebook is run from."""
    for folder in [Path("datasets"), Path("../datasets")]:   # the folder is here, or one level up
        if (folder / name).exists():                          # is the file in this folder?
            return folder / name                              # yes: return its path
    raise FileNotFoundError(
        f"Could not find {name}. Open this notebook from inside the course folder, and make sure "
        "the course was unpacked (on Windows: right-click the ZIP > Extract All).")

df = pd.read_csv(find_dataset(DATASET_FILE))   # open the CSV file as a table
print("Records:", len(df))                     # how many rows (records) the table has
```

```python
#@title Solution (unfold to see the code)
providers = df[PROVIDER_COLUMN].value_counts()         # records per provider, largest first
percent = (providers / len(df) * 100).round(1)          # the same as a percentage of all records

table = pd.DataFrame({"records": providers, "percent": percent})   # put both into one table
table
```

### 4 · From which countries?

The same count for countries, drawn as a horizontal bar chart (horizontal bars leave room for long
names).

#### Your turn: from which countries?

**Task:** count the records per country and draw them as a horizontal bar chart.

**Hint:** `.value_counts()` gives the counts; `.plot(kind="barh")` draws horizontal bars.

**You should get:** nine countries, with Germany's bar far longer than all the others.

Try it in the empty cell below. Then unfold the **solution** cell under it, compare, and run the solution
cell too, because the next steps use it.

```python
# Your attempt: write your code here, then press Shift + Enter
```

```python
#@title Solution (unfold to see the code)
countries = df[COUNTRY_COLUMN].value_counts()          # records per country
print(countries)

ordered = countries.sort_values()                       # smallest first, so the largest bar is on top
ax = ordered.plot(kind="barh", figsize=(8, 4), color="#2E75B6")   # horizontal bar chart
ax.set_title("Where do the records come from?")         # chart title
ax.set_xlabel("Number of records")                      # label of the horizontal axis
ax.set_ylabel("")                                       # no label needed on the vertical axis
plt.tight_layout()
plt.show()
```
