2. Findable: discoverable through metadata and a persistent identifier. Accessible: clear how to obtain it and under which conditions. Interoperable: connects to other data through shared formats, identifiers, or vocabularies. Reusable: its content and conditions of use are clear enough for further research.
3. No. Data can need controlled access because of copyright, privacy, or ethics. FAIR then requires that the access conditions and the procedure for requesting access are clearly stated.
4. A Digital Object Identifier: a persistent identifier for one specific published output or version, which supports citation and links related outputs. On its own it is not enough: the dataset also needs metadata, a description, a version and date, access conditions, and a licence.

6. Share as much as you can, but restrict what must be restricted (copyright, privacy, contracts, community rights); usually you can still share metadata and a description. For Indigenous or community-governed data, the CARE principles add collective benefit, authority to control, responsibility, and ethics.

### 3.0 · Before you start: practise first

Publishing on Zenodo is **permanent**: a DOI cannot be deleted (you can only publish new versions). So
practise first on the **Zenodo Sandbox**, <https://sandbox.zenodo.org>, a copy of Zenodo where nothing is
permanent and DOIs are only for testing. It needs its own account (you can sign in with GitHub or ORCID).

Publish on the real Zenodo only a dataset that **you created**. If you worked with the course's Van Gogh
data, publish your practice record on the Sandbox only: the course dataset does not need to be published
again by every participant.

### 3.1 · Find your gaps with a FAIR self-assessment

1. Open the ARDC [FAIR Data Self-Assessment Tool](https://ardc.edu.au/resource/fair-data-self-assessment-tool/).
2. Answer the questions for your dataset as it is now: is there a persistent identifier? rich
 metadata? a clear licence?
3. Note the items where you score low. Usually these are *no persistent identifier* and *no licence*.
 Publishing on Zenodo fixes most of them.

### 3.2 · Decide what to publish

Publish **what you created, not what you downloaded**: your filtered, cleaned, or enriched version, with
the fields you added. If your data is just the original, link to the original instead. One project becomes
one Zenodo record with one DOI; several related files can go into that single record.

### 3.3 · Decide how to release, and prepare the files

**First, decide.** Releasing openly is not the only responsible choice. In the configuration cell of the
notebook, set `RELEASE_DECISION` to one of:

- `"open"`: anyone may access and reuse the data under your licence (the usual case for the course data);
- `"restricted"`: the description is public, but the files are shared only on request (for example because
 of copyright or personal data);
- `"not released"`: you do not publish the data, for example because the rights do not allow it, or the
 community it concerns has not agreed (the CARE principles).

Write who may access the data, and how, in `ACCESS_STATEMENT`, and your reasons in `DECISION_REASON`. A
justified decision not to release is a valid result of this module.

**Then prepare the files:**

1. Open **08-open-publishing > 08_prepare_publication.ipynb**.
2. In the configuration cell, fill in the title, creators (your name, and your ORCID if you have one),
 a description of what the data is and how you made it, keywords, the licence, and the source.
3. Run the notebook. It tells you whether your dataset differs from the original, lists anything still
 missing, and builds a folder `publication` containing the data file (unless you decided not to release
 it), a `README.md` that describes every column and the access conditions, a `CITATION.cff` citation file,
 `zenodo_form.txt` with the text for each field of the Zenodo form, and `release_decision.md`, your
 documented decision. It also copies the package into your project.

If you chose *restricted*, you can still publish on Zenodo: describe the dataset openly and set the files
to restricted access, so that people must ask. If you chose *not released*, skip 3.4 to 3.6, and keep the
release decision in your project.

### 3.4 · Start an upload on Zenodo

1. Go to <https://sandbox.zenodo.org> (practice) or <https://zenodo.org> (real), and sign up or log in.
2. Click **New upload** (or the **+** at the top, then *New upload*).
3. Drag the files from your `publication` folder into the upload area: the data file and the `README.md`.
 Wait until each file shows as uploaded.

### 3.5 · Fill in the metadata

Open `zenodo_form.txt` from the notebook and copy each part into the form:

- **Do you already have a DOI?** Choose **No**, so that Zenodo creates one.
- **Resource type:** *Dataset*.
- **Title**, **Publication date** (today), and **Creators** (click *Add creator*; add your ORCID if you have one).
- **Description:** what the data is, where it comes from, and how you made it.
- **Licenses:** choose the licence you set in the notebook, for example *Creative Commons Attribution 4.0
 International* (CC BY 4.0) or *Creative Commons Zero* (CC0).
- **Keywords**, and **Version** (for example `1.0`).
- **Related works:** add your GitHub repository's address, so that the data and the code point at each other.

The richer the metadata, the easier your data is to find.

### 3.6 · Publish and get your DOI

1. Click **Preview** and check everything.
2. Click **Publish**, and confirm.
3. The DOI appears on the record page. On the real Zenodo it is permanent.

### 3.7 · Check again, and close the loop

1. Run the FAIR self-assessment once more. *Findable* and *Reusable* should now score higher, because you
 added an identifier, metadata, and a licence.
2. Add the DOI to your project: put the link in your GitHub README, add it to `CITATION.cff`, then commit
 and push. Now your repository, your code, and your published dataset all point at each other.

### Configuration

Describe your dataset here. The values below are filled in for the course example; replace them for your
own dataset. Everything you write here ends up in the README and on Zenodo.

```python
DATASET_FILE  = "van_gogh_enhanced.csv"   # the dataset you created (falls back to the filtered one)
ORIGINAL_FILE = "van_gogh_combined.csv"   # the data as you obtained it, to compare with

TITLE    = "Works catalogued under Vincent van Gogh's name on Europeana, filtered and linked to Wikidata"
CREATORS = [("Surname, Given name", "")]  # one entry per creator: ("Surname, Given name", "ORCID or empty")
DESCRIPTION = (
    "Records of works catalogued under several name forms of Vincent van Gogh on Europeana, "
    "combined and de-duplicated, with a photographic reproduction index set aside so that the data "
    "reflect holding institutions. Each institution is linked to its Wikidata identifier and country. "
    "Created as a teaching example in the course Open Research in the Digital Humanities."
)
KEYWORDS = ["Van Gogh", "Europeana", "cultural heritage", "open data", "Wikidata"]
LICENSE  = "CC-BY-4.0"                    # for example "CC-BY-4.0" or "CC0-1.0"
VERSION  = "1.0"

SOURCE_NAME    = "Europeana Search API"
SOURCE_URL     = "https://www.europeana.eu"
SOURCE_LICENSE = "CC0 1.0 (Europeana metadata)"
ACCESS_DATE    = "2026-09-15"             # when you obtained the original data
GITHUB_URL     = ""                       # the address of your project repository on GitHub

# Your release decision: "open", "restricted", or "not released".
# Restricted or not released is a valid, responsible choice, if you say why.
RELEASE_DECISION = "open"
ACCESS_STATEMENT = "Openly available to anyone under the licence above."     # who may access it, and how
DECISION_REASON  = ("The dataset is derived from Europeana metadata published under CC0 and contains "
                    "no personal data, so it can be shared openly.")

# Your own project from Module 2. Results are also copied there, if the folder exists.
from pathlib import Path
PROJECT_FOLDER = Path.home() / "Documents" / "GitHub" / "open-project-2026"

OUTPUT_FOLDER = "publication"             # the folder the package is built in
```

### 1 · Open your dataset and the original

```python
from pathlib import Path        # a safe way to work with file paths
import pandas as pd             # the library for tables

def find_dataset(name):
    """Return the path to a file in the course's datasets folder, wherever the notebook is run from."""
    for folder in [Path("datasets"), Path("../datasets")]:   # the folder is here, or one level up
        if (folder / name).exists():                          # is the file in this folder?
            return folder / name                              # yes: return its path
    raise FileNotFoundError(
        f"Could not find {name}. Open this notebook from inside the course folder, and make sure "
        "the course was unpacked (on Windows: right-click the ZIP > Extract All).")

try:
    data_path = find_dataset(DATASET_FILE)                  # the dataset you created
except FileNotFoundError:                                   # Module 6 is optional: fall back
    data_path = find_dataset(DATASET_FILE.replace("_enhanced", "_filtered"))
    print("Note: using", data_path.name, "because", DATASET_FILE, "was not found.")

data = pd.read_csv(data_path)                               # your dataset
original = pd.read_csv(find_dataset(ORIGINAL_FILE))         # the original, for comparison
print(f"Your dataset: {data_path.name}, {len(data)} records, {len(data.columns)} fields")
print(f"Original:     {ORIGINAL_FILE}, {len(original)} records, {len(original.columns)} fields")
```

### 2 · Did you create something new?

Publish a separate dataset only if you changed the original in a meaningful way: filtered, corrected,
standardised, enriched, restructured, combined, annotated, or classified it. Otherwise, cite and link the
original instead.

```python
added   = [c for c in data.columns if c not in original.columns]     # fields you added
removed = [c for c in original.columns if c not in data.columns]     # fields you removed

CHANGES = []                                                          # a list of what you changed
if len(data) < len(original):
    CHANGES.append(f"filtered from {len(original)} to {len(data)} records")
if added:
    CHANGES.append("added the fields " + ", ".join(added))
if removed:
    CHANGES.append("removed the fields " + ", ".join(removed))

if CHANGES:
    print("[OK] Your dataset differs from the original:")
    for c in CHANGES:
        print("   -", c)
    print("Publishing it as a new dataset is appropriate.")
else:
    print("[CHECK] Your dataset looks the same as the original.")
    print("If you did not change it, cite and link the original instead of publishing a copy.")
```

### 4 · Build the publication folder

This cell creates the folder `publication`: the result of Module 8. It contains a copy of your data
(unless you decided not to release it), a `README.md` that describes the dataset, every column, and the
access conditions, a `CITATION.cff` file that tells people how to cite it, `zenodo_form.txt` with the text
for the Zenodo form, and `release_decision.md`, your documented release decision.

```python
import datetime
import shutil                                                       # to copy files

out = Path(OUTPUT_FOLDER)
if out.exists():                                                    # start from an empty folder,
    shutil.rmtree(out)                                              # so no old files are left in it
out.mkdir()                                                         # create the folder
if RELEASE_DECISION != "not released":                              # 1) a copy of the data file
    shutil.copy(data_path, out / data_path.name)                    #    (not if you decided not to release it)
TODAY = datetime.date.today().isoformat()

def kind(series):
    """A plain-language type for a column: number or text."""
    if series.isna().all():
        return "empty"
    return "number" if pd.api.types.is_numeric_dtype(series) else "text"

def example(series):
    """The first non-empty value of a column, shortened."""
    values = series.dropna()
    return str(values.iloc[0])[:40].replace("|", "/") if len(values) else ""

# 2) README.md: a datasheet describing the dataset
rows = [f"| `{c}` | {kind(data[c])} | {example(data[c])} | {data[c].isna().mean():.0%} |" for c in data.columns]
creators_text = "; ".join(name + (f" ({orcid})" if orcid else "") for name, orcid in CREATORS)
sparse = [c for c in data.columns if data[c].isna().mean() > 0.5]
readme = f"""# {TITLE}

**Creators:** {creators_text} · **Version:** {VERSION} · **Licence:** {LICENSE} · **Date:** {TODAY}

## Description
{DESCRIPTION}

## Source
- Original data: {SOURCE_NAME} ({SOURCE_URL})
- Licence of the original: {SOURCE_LICENSE}
- Accessed: {ACCESS_DATE}

## How this dataset was made
Compared with the original data, this dataset was {'; '.join(CHANGES) if CHANGES else 'not changed'}.
Code and documentation: {GITHUB_URL or '(add the link to your GitHub repository)'}

## Fields
{len(data)} records, {len(data.columns)} fields.

| Field | Type | Example | Missing |
|---|---|---|---|
""" + "\n".join(rows) + f"""

## Known limitations
{('Mostly empty fields: ' + ', '.join(sparse) + '. ') if sparse else ''}Add any other limitations you know of here.

## Access and release
- **Release decision:** {RELEASE_DECISION}
- **Access:** {ACCESS_STATEMENT}
- **Reason:** {DECISION_REASON}

## How to cite
{creators_text} ({TODAY[:4]}). *{TITLE}* (version {VERSION}). The DOI will be added after publication.
"""
(out / "README.md").write_text(readme, encoding="utf-8")

# 3) CITATION.cff: a small, standard file that tells people how to cite the dataset
authors = []
for name, orcid in CREATORS:
    family, _, given = name.partition(",")                         # split "Surname, Given name"
    authors.append(f'  - family-names: "{family.strip()}"\n    given-names: "{given.strip()}"'
                   + (f'\n    orcid: "{orcid}"' if orcid else ""))
cff = f"""cff-version: 1.2.0
message: "If you use this dataset, please cite it as below."
type: dataset
title: "{TITLE}"
version: "{VERSION}"
date-released: "{TODAY}"
license: "{LICENSE}"
authors:
""" + "\n".join(authors) + "\nkeywords:\n" + "\n".join(f'  - "{k}"' for k in KEYWORDS) + "\n"
if GITHUB_URL:
    cff += f'repository-code: "{GITHUB_URL}"\n'
(out / "CITATION.cff").write_text(cff, encoding="utf-8")

# 4) zenodo_form.txt: what to paste into each field of the Zenodo form
form = f"""Do you already have a DOI?   No
Resource type:               Dataset
Title:                       {TITLE}
Publication date:            {TODAY}
Creators:                    {creators_text}
Description:                 {DESCRIPTION}
Licenses:                    {LICENSE}
Keywords:                    {', '.join(KEYWORDS)}
Version:                     {VERSION}
Related works:               {GITHUB_URL or '(your GitHub repository address)'}
Access:                      {"Open" if RELEASE_DECISION == "open" else "Restricted (files only on request)" if RELEASE_DECISION == "restricted" else "Do not upload: decided not to release"}
Files to upload:             {data_path.name + ", README.md" if RELEASE_DECISION != "not released" else "none (keep the documentation in your project)"}
"""
(out / "zenodo_form.txt").write_text(form, encoding="utf-8")

# 5) release_decision.md: the documented release decision
decision = f"""# Release decision

*Recorded {TODAY} with the course notebook 08_prepare_publication.*

- **Dataset:** {TITLE} (version {VERSION})
- **Decision:** {RELEASE_DECISION} (options considered: open, restricted, not released)
- **Access statement:** {ACCESS_STATEMENT}
- **Reason:** {DECISION_REASON}
- **Licence:** {LICENSE}; licence of the source data: {SOURCE_LICENSE}
- **What is shared:** {"the data file, README, and citation file" if RELEASE_DECISION == "open" else "the description and metadata; the data file only on request" if RELEASE_DECISION == "restricted" else "nothing yet; the documentation stays in the project"}
"""
(out / "release_decision.md").write_text(decision, encoding="utf-8")

for f in sorted(out.iterdir()):
    print("[CREATED]", f)
```
